## - Emissions Data from EDGAR
### Data Importation

```python
import pandas as pd

# Initializing the file path
file_path = "EDGAR_2024_GHGs_CO2eq_AR5_NUTS2_by_country_sector_1990-2023_v3.xlsx"

# List of Sheets to import
sheets_to_load = ["Fossil CO2", "CH4 AR5", "N2O AR5", "F-gases AR5"]

# Skipping the first 6 rows since column names start in row 7
dfs = pd.read_excel(
    file_path,
    sheet_name=sheets_to_load,
    skiprows=6  
)

# Getting the data on emissions 
df_Fossil_CO2 = dfs["Fossil CO2"]
df_CH4_AR5 = dfs["CH4 AR5"]
df_N2O_AR5 = dfs["N2O AR5"]
df_F_gases_AR5 = dfs["F-gases AR5"]

```

### Data Cleaning, Transformation and Feature Engineering

```python
# Filtering each dataset for USA and Industry Sector
dfs1 = [df_Fossil_CO2, df_CH4_AR5, df_N2O_AR5, df_F_gases_AR5]
df_names = ["Fossil_CO2", "CH4_AR5", "N2O_AR5", "F_gases_AR5"]

# Initializing the filtered dataset
filtered_dfs = []

# Lopping over each dataset
for df, name in zip(dfs1, df_names):
    temp = df.copy()

    # Filtering for United States and Industry sector
    temp = temp[(temp["Country"] == "United States") & (temp["Sector"] == "Industry")]

    # Reshaping from wide to long 
    temp = temp.melt(
        id_vars=["Country", "Subnational desc"],
        value_vars=[col for col in temp.columns if col.startswith("Y_")],
        var_name="Year",
        value_name="Emission"
    )

    # Renaming columns
    temp["Year"] = temp["Year"].str.replace("Y_", "").astype(int)
    temp = temp.rename(columns={"Subnational desc": "State"})
    temp["Emission_Type"] = name

    # Reordering columns
    temp = temp[["Country", "State", "Year", "Emission_Type", "Emission"]]
    filtered_dfs.append(temp)

# Combining all datasets
df_combined = pd.concat(filtered_dfs, ignore_index=True)

# Aggregating by State and Year 
df_agg = (
    df_combined.groupby(["Country", "State", "Year"], as_index=False)
    .agg({"Emission": "sum"})
)

```

## - GDP Data from FRED
### Data Importation

```python
# Initializing the file path
file_path1 = "GDP_per_state_Q1_1990_to_Q2_2023.xlsx"

# List of Sheets to import
sheets_to_load1 = ["Quarterly"]

# Importing Data
dfs = pd.read_excel(
    file_path1,
    sheet_name=sheets_to_load1 
)

# Getting the data on emissions 
df_Quarterly = dfs["Quarterly"]
```

### Data Overview

```python
print(df_Quarterly.head())
```

### Data Cleaning, Transformation and Feature Engineering

```python
#  Converting from wide format to long format
df_Quarterly_long = df_Quarterly.melt(
    id_vars=["observation_date"],
    var_name="State_Code",
    value_name="Manufacturing_GDP"
)

df_Quarterly_long.head()
```

```python
# Extracting the 2-letter state abbreviation 
df_Quarterly_long["State"] = df_Quarterly_long["State_Code"].str[:2]
df_Quarterly_long.head()
```

```python
# Mapping the state abbreviations for full state names
us_states = {
    'AK': 'Alaska', 'AL': 'Alabama', 'AR': 'Arkansas', 'AZ': 'Arizona', 'CA': 'California',
    'CO': 'Colorado', 'CT': 'Connecticut', 'DC': 'District of Columbia', 'DE': 'Delaware',
    'FL': 'Florida', 'GA': 'Georgia', 'HI': 'Hawaii', 'IA': 'Iowa', 'ID': 'Idaho',
    'IL': 'Illinois', 'IN': 'Indiana', 'KS': 'Kansas', 'KY': 'Kentucky', 'LA': 'Louisiana',
    'MA': 'Massachusetts', 'MD': 'Maryland', 'ME': 'Maine', 'MI': 'Michigan', 'MN': 'Minnesota',
    'MO': 'Missouri', 'MS': 'Mississippi', 'MT': 'Montana', 'NC': 'North Carolina',
    'ND': 'North Dakota', 'NE': 'Nebraska', 'NH': 'New Hampshire', 'NJ': 'New Jersey',
    'NM': 'New Mexico', 'NV': 'Nevada', 'NY': 'New York', 'OH': 'Ohio', 'OK': 'Oklahoma',
    'OR': 'Oregon', 'PA': 'Pennsylvania', 'RI': 'Rhode Island', 'SC': 'South Carolina',
    'SD': 'South Dakota', 'TN': 'Tennessee', 'TX': 'Texas', 'UT': 'Utah', 'VA': 'Virginia',
    'VT': 'Vermont', 'WA': 'Washington', 'WI': 'Wisconsin', 'WV': 'West Virginia',
    'WY': 'Wyoming'
}

df_Quarterly_long["State_Name"] = df_Quarterly_long["State"].map(us_states)
df_Quarterly_long.head()
```

```python
# Cleanin up and reordering columns ---
df_Quarterly_long = df_Quarterly_long.rename(columns={"observation_date": "Date"})
df_Quarterly_long = df_Quarterly_long[["Date", "State_Name", "Manufacturing_GDP"]]
df_Quarterly_long.head()
```

```python
# Convert Date to datetime 
df_Quarterly_long["Date"] = pd.to_datetime(df_Quarterly_long["Date"])
# Creating year variable
df_Quarterly_long["Year"] = df_Quarterly_long["Date"].dt.year
df_Quarterly_long.head()
```

```python
# Aggregating over yearly data
df_yearly = df_Quarterly_long.groupby(["Year", "State_Name"], as_index=False)["Manufacturing_GDP"].sum()
df_yearly.head()
```

```python
# Checking data info
df_yearly.info()
```

```python
#Checking missing data
df_yearly.isnull().sum()
```
